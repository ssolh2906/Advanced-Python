```python
from sqlalchemy.orm import declarative_base
```

```python
Base = declarative_base()

# Book table
class Book(Base):
    __tablename__ = "book"
    book_id = Column(Integer, primary_key=True)
    title = Column(String, nullable=False)
    price = Column(Float)
    borrowitems = relationship("BorrowItem", back_populates="book")
    

# Member table
class Member(Base):
    __tablename__ = "member"
    member_id = Column(Integer, primary_key=True)
    name = Column(String, nullable=False)
    email = Column(String, nullable=False)
    borrows = relationship("Borrow", back_populates="member")
    

# Borrow table
class Borrow(Base):
    __tablename__= "borrow"
    borrow_id = Column(Integer, primary_key=True)
    member_id = Column(Integer, ForeignKey('member.member_id'))
    borrow_date = Column(String)
    member = relationship("Member", back_populates="borrows")
    borrowitems = relationship("BorrowItem", back_populates=True)

# Borrow item
class BorrowItem(Base):
    __tablename__="borrowitem"
    borrow_id = Column(Integer, ForeignKey('borrow.borrow_id'), primary_key=True)
    book_id = Column(Integer, ForeignKey('book.book_id'), primary_key=True)
    quantity = Column(Integer)
    book = relationship("Book", back_populates="borrowitems")
    borrow = relationship("Borrow", back_populates="borrowitems")
    
```
